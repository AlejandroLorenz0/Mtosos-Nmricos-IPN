```python
import math 
#la función que queremos integrar es f(x)# 
def f(x):
    return x**2
```

```python
#primero definimos las entradas#
a=float(input("Ingrese el valor del limite inferior (a): "))
b=float(input("Ingrese el valor del limite superior (b): "))
n=int(input("ingrese el numero de intervalos (n debe ser par): "))
#podemos agregar una validación para regresar en caso de ingresar un numero impar en la variable n#
if n % 2 !=0:
    print("el numero de n debe ser par")
```

```python
#definimos ahora a h#
h = (b - a) / n
#inicializamos las sumas#
#XIO es la suma de los extremos, XI1 es la suma de los puntos impares, XI2 es la suma de los puntos pares y ambos inician en 0#
XI0 = f(a) + f(b)
XI1 = 0.0
XI2 = 0.0
#ahora generamos el bucle principal del método#
for i in range(1, n):
    #calculamos x aqui#
    x= a+ i* h
    #para la condición impar#
    if i % 2 == 0:
        XI2= XI2 + f(x)
    #para la condición par#
    else:
        XI1= XI1 + f(x)
    
```

```python
#para el cálculo final#
XI = h * (XI0 + 2 * XI2 + 4 * XI1)/3
#finalmente la salida del procedimiento#
print(f"el valor aproximado de la integral es: {XI}")
```
