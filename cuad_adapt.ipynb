**CUADRATURA ADAPTABLE**

```python
import numpy as np 
##definimos primero la función que queremos integrar##
def f(x):
    return (100.0 / (x**2)) * np.sin(10.0 / x)
#ahora definimos los parametros de entrada del algoritmo#
a = 1.0           #extremo inferior
b = 3.0           #extremo superior
tol = 1e-4        #tolerancia
n = 20           #cota máxima de niveles
```

```python
def cuadratura(a, b, tol, n, f):
    # Inicializamos las variables para las pilas de datos
    max_size = 1000
    v1 = np.zeros(max_size)
    v2 = np.zeros(max_size)
    v3 = np.zeros(max_size)
    v4 = np.zeros(max_size)
    v5 = np.zeros(max_size)
    v6 = np.zeros(max_size)
    v7 = np.zeros(max_size)
    v8 = np.zeros(max_size, dtype=int)
    
    # Inicializamos las variables globales y del primer nivel
    APP = 0.0
    i = 1
    TOL_i = 10.0 * tol
    a_i = a
    h_i = (b - a) / 2.0
    FA_i = f(a_i)
    FC_i = f(a_i + h_i)
    FB_i = f(b)
    
    # Aproximación inicial a partir del método de Simpson para el intervalo completo
    S_i = h_i * (FA_i + 4.0 * FC_i + FB_i) / 3.0
    L_i = 1
    
    # Iniciamos un bucle mientras existan niveles pendientes
    while i > 0:
        # Evaluar puntos medios y aproximar con Simpson para mitades
        x_fd = a_i + h_i / 2.0
        x_fe = a_i + 3.0 * h_i / 2.0
        
        FD = f(x_fd)
        FE = f(x_fe)
        
        S1 = h_i * (FA_i + 4.0 * FD + FC_i) / 6.0
        S2 = h_i * (FC_i + 4.0 * FE + FB_i) / 6.0
        
        # Guardamos los datos actuales en el nivel i
        v1[i] = a_i
        v2[i] = FA_i
        v3[i] = FC_i
        v4[i] = FB_i
        v5[i] = h_i
        v6[i] = TOL_i
        v7[i] = S_i
        v8[i] = L_i
        
        # Eliminamos temporalmente el nivel actual de la pila para evaluar
        i = i - 1
        
        # Comprobar si la precisión es aceptable
        if abs(S1 + S2 - v7[i+1]) < v6[i+1]:
            # Si se cumple acumulamos la aproximación
            APP = APP + (S1 + S2)
        else:
            # Si no se cumple verificamos si se supera el nivel máximo n
            if v8[i+1] >= n:
                return None
            else:
                # Añadimos un subintervalo a la derecha
                i = i + 1
                v1[i] = v1[i-1] + v5[i-1]
                FA_i = v3[i-1]
                FC_i = FE
                FB_i = v4[i-1]
                h_i = v5[i-1] / 2.0
                TOL_i = v6[i-1] / 2.0
                S_i = S2
                L_i = v8[i-1] + 1
                
                # Añadimos un subintervalo a la izquierda
                i = i + 1
                v1[i] = v1[i-1]
                FA_i = v2[i-1]
                FC_i = FD
                FB_i = v3[i-1]
                h_i = v5[i-2] / 2.0
                TOL_i = v6[i-2] / 2.0
                S_i = S1
                L_i = v8[i-2] + 1
                
    # Retornamos el valor final de la aproximación APP
    return APP
```

```python
# Definimos los parámetros de prueba según el libro
a = 1.0
b = 3.0
tol = 1e-4
n = 20

# Llamamos a la función 'cuadratura' que definimos en el bloque anterior
resultado = cuadratura(a, b, tol, n, f)

if resultado is not None:
    print(f"Aproximación de la integral (APP): {resultado:.6f}")
```
